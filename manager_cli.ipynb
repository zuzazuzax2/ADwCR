```python
# import json
import os


CONTROL_STATE_PATH = "/home/jovyan/notebooks/control_state.json"

INGREDIENTS = [
    "pizza_dough",
    "tomato_sauce",
    "cheese",
    "pepperoni",
    "bun",
    "beef_patty",
    "lettuce",
    "pasta",
    "bacon",
    "fries",
    "chicken",
    "caesar_sauce",
    "soup_base"
]

DEFAULT_CONTROL_STATE = {
    "current_staff": 3,
    "manual_restock": {}
}


def load_control_state():
    if not os.path.exists(CONTROL_STATE_PATH):
        save_control_state(DEFAULT_CONTROL_STATE)
        return DEFAULT_CONTROL_STATE.copy()

    with open(CONTROL_STATE_PATH, "r", encoding="utf-8") as file:
        return json.load(file)


def save_control_state(state):
    with open(CONTROL_STATE_PATH, "w", encoding="utf-8") as file:
        json.dump(state, file, ensure_ascii=False, indent=2)


def show_state():
    state = load_control_state()

    print("\n=== AKTUALNY STAN DECYZJI MENADŻERA ===")
    print(f"Liczba pracowników: {state.get('current_staff', 3)}")
    print("Oczekujące uzupełnienie składników:")

    manual_restock = state.get("manual_restock", {})

    if manual_restock:
        for ingredient, amount in manual_restock.items():
            print(f"  - {ingredient}: +{amount}")
    else:
        print("  brak")

    print("=======================================\n")


def change_staff():
    state = load_control_state()

    try:
        new_staff = int(input("Podaj nową liczbę pracowników, np. 4: "))

        if new_staff < 1:
            print("Liczba pracowników musi być większa od 0.")
            return

        state["current_staff"] = new_staff
        save_control_state(state)

        print(f"Zmieniono liczbę pracowników na: {new_staff}")

    except ValueError:
        print("Podaj liczbę całkowitą.")


def add_ingredient():
    state = load_control_state()

    print("\nDostępne składniki:")

    for index, ingredient in enumerate(INGREDIENTS, start=1):
        print(f"{index}. {ingredient}")

    try:
        choice = int(input("Wybierz numer składnika: "))

        if choice < 1 or choice > len(INGREDIENTS):
            print("Niepoprawny numer składnika.")
            return

        ingredient = INGREDIENTS[choice - 1]

        amount = int(input(f"Ile dodać składnika {ingredient}? "))

        if amount <= 0:
            print("Ilość musi być większa od 0.")
            return

        manual_restock = state.get("manual_restock", {})
        manual_restock[ingredient] = manual_restock.get(ingredient, 0) + amount

        state["manual_restock"] = manual_restock
        save_control_state(state)

        print(f"Dodano do kolejki uzupełnienia: {ingredient} +{amount}")

    except ValueError:
        print("Podaj poprawną liczbę.")


def reset_manual_restock():
    state = load_control_state()
    state["manual_restock"] = {}
    save_control_state(state)

    print("Wyczyszczono oczekujące uzupełnienia składników.")


def main():
    while True:
        print("\n=== PANEL MENADŻERA RESTAURACJI ===")
        print("1. Pokaż aktualny stan")
        print("2. Zmień liczbę pracowników")
        print("3. Dodaj składnik do spiżarni")
        print("4. Wyczyść oczekujące uzupełnienia")
        print("5. Wyjdź")

        choice = input("Wybierz opcję: ")

        if choice == "1":
            show_state()

        elif choice == "2":
            change_staff()

        elif choice == "3":
            add_ingredient()

        elif choice == "4":
            reset_manual_restock()

        elif choice == "5":
            print("Zamykam panel menadżera.")
            break

        else:
            print("Niepoprawna opcja. Wybierz 1–5.")


if __name__ == "__main__":
    main()
```
